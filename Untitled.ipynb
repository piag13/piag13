```python
import numpy as np

class SoftmaxRegression:
    def __init__(self, lr = 0.01, epochs = 1000):
        self.lr = lr
        self.epochs = epochs
        self.w = []
        self.b = []
        self.loss = []
        
    def softmax(self, z):
        exp_z = np.exp(z)
        sum_exp_z = np.sum(exp_z, axis = 1, keepdims = True)
        return exp_z / sum_exp_z
    
    def cross_entropy(self, y, y_pred):
        return -np.sum(y * np.log(y_pred)) / y.shape[0]
    
    def predict(self, X):
        z = np.dot(X, self.w) + self.b
        y_pred = self.softmax(z)
        return y_pred

    def onehot_encode(self, y):
        number_labels = len(y)
        unique_labels = len(np.unique(y))
        y_one_hot = np.zeros((number_labels, unique_labels))
        y_one_hot[np.arange(numbers_labels), y - 1] = 1
        return y_one_hot


    def fit(self, X, y):
        indices = np.arange(X.shape[0])
        np.random.shuffle(indices)
        X = X[indices]
        y = y[indices]

        num_samples, num_features = X.shape
        num_class = len(np.unique(y))

        y_onehot = self.onehot_encode(y)

        #intitalize parameter
        self.w = np.random.randn(num_features, num_class) * 0.01
        self.b = np.zeros((1, num_class))

        #training
        for epoch in range(self.num_epochs):
            y_pred = self.predict(X)
            loss = self.cross_entropy(y_onehot, y_pred)
            self.losses.append(loss)

            dz = y_pred - y_onehot
            dw = np.dot(X.T, dz) / num_samples
            db = np.sum(dz, axis = 0, keepdims = True)/num_samples

            self.w = self.w - self.lr * dw
            self.b = self.b - self.lr * db

            if epoch % 100 == 0:
                print(f"Epoch: {epoch}, loss: {loss}")
```
