```python
import numpy as np
import pandas as pd

class SoftmaxRegression:
    def __init__(self, learning_rate = 0.01, epochs = 1000):
        self.learning_rate = learning_rate
        self.epochs = epochs
        self.W = None
        self.b = None
        
    def predict(self, X):
        z = np.dot(X.T, self.W) + self.b
        y_pred = self.softmax(z)    
        return np.argmax(y_pred, axis = 1)

    def softmax(self, z):
        exp_z = np.exp(z)
        sum_exp = np.sum(exp_z, axis = 1, keepdims = True)
        return exp_z/sum_exp
        
    def cross_entropy(self, y, y_pred):
        loss = -np.sum(y * np.log(y_pred))/y.shape[0]
        return loss

    def fit(self, X, y):
        n_samples, n_features = X.shape
        self.b = np.zeros((1, np.max(y) + 1))
        self.W = np.zeros((n_features, np.max(y) + 1))
        for i in range(self.epochs):
            logits = np.dot(X.T, self.W) + self.b
            y_pred = self.softmax(logits)
            
```
